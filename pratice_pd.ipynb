```python
df["Parents"] = df["SibSp"] + df["Parch"]
df
```

```python
df.groupby("Parents")["Survived"].mean().sort_values(ascending=False).head(1)
```

```python
df['FareGroup'] = pd.cut(df['Fare'],
                         bins=[0, 25, 50, 100, 250],
                         labels=['Low', 'Medium', 'High', 'Very High'])

df.groupby('FareGroup')['Survived'].mean() * 100
```

```python
df.groupby("Pclass")["Survived"].mean()
```

```python
df.groupby("Pclass")["Survived"].agg(
    SurvivalRate="mean",
    PassengerCount="count"
).sort_values(by="SurvivalRate", ascending=False)
```

```python
df['AgeGroup'] = pd.cut(
    df['Age'],
    bins=[0, 12, 50, 100],
    labels=['Child', 'Adult', 'Senior']
)
df[['Age', 'AgeGroup']]
df.groupby('AgeGroup')['Survived'].mean()
```
