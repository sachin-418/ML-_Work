```python
import pandas as pd
import numpy as np

df = pd.read_csv("Dataset 1_origin.csv")
df.sample(20)
```

```python
df = df.loc[: , 'Education':'Kidhome']
df = df.drop('Income',axis=1)
```

```python
df
```

```python
df["Marital_Status"].value_counts()
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(df.drop("Kidhome", axis=1), df["Kidhome"], test_size=0.25 , random_state=0)
```

```python
from sklearn.preprocessing import OrdinalEncoder
oe=OrdinalEncoder(categories=[["PhD","Master","Graduation","2n Cycle","Basic"],["Married","Together","Divorced","Widow","Single","Alone","Absurd","YOLO"]])
```

```python
oe.fit(x_train)
```

```python
x_train = oe.transform(x_train)
x_test = oe.transform(x_test)
```
