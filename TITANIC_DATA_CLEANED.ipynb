```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("tested.csv")
df.sample(10)
```

```python
df.isnull().mean()*100
```

```python
df.drop(columns=["Cabin","PassengerId","Name","Ticket","Fare"], inplace=True)
df["Family"] = df["SibSp"] + df["Parch"]
df.drop(columns=["SibSp","Parch"], inplace=True)
```

```python

df["Family"].value_counts()
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test , y_train , y_test = train_test_split(df.drop("Survived",axis = 1), df["Survived"], test_size=0.2, random_state=42)
```

```python
x_train
```

```python
y_train
```

```python
cle = ColumnTransformer(transformers=[
    ("age_imp", SimpleImputer(strategy='mean'), ["Age"]),
    ("sex_emb", OneHotEncoder(drop="first", sparse_output=False), ["Sex","Embarked"]),

    ("scale", StandardScaler(), ["Age","Family"]),

], remainder="passthrough")
```

```python
cle.fit_transform(x_train).shape
```

```python
dfc = DecisionTreeClassifier()
dfc.fit(cle.fit_transform(x_train), y_train)
```

```python
y_pred = dfc.predict(cle.transform(x_test))
```

```python
y_pred
```

```python
accuracy = accuracy_score(y_test, y_pred)
accuracy
```
