```python
# --- Cell 1 ---
from google.colab import files
uploaded = files.upload()

# --- Cell 2 ---
import pandas as pd
import numpy as np
df = pd.read_csv("covid_toy.csv")
df

# --- Cell 3 ---
df["city"].value_counts() , df["gender"].value_counts()

# --- Cell 4 ---
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    df.drop("has_covid", axis=1),
    df["has_covid"],
    test_size=0.2,
    random_state=0
)

# --- Cell 5 ---
from sklearn.impute import SimpleImputer

si = SimpleImputer()

x_train_fever = si.fit_transform(x_train[["fever"]])
x_test_fever = si.transform(x_test[["fever"]])

# --- Cell 6 ---
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories=[["Mild", "Strong"]] , dtype=np.int32)

x_train_cough = oe.fit_transform(x_train[["cough"]])
x_test_cough = oe.transform(x_test[["cough"]])

# --- Cell 7 ---
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(drop="first", sparse_output=False, dtype= np.int32)

x_train_gender_city= ohe.fit_transform(x_train[["gender", "city"]])
x_test_gender_city = ohe.transform(x_test[["gender", "city"]])

x_test_gender_city.shape

# --- Cell 8 ---
x_train_age = x_train.drop(columns=["gender", "fever", "cough", "city"]).values
x_test_age = x_test.drop(columns= ["gender","fever","cough","city"]).values

# --- Cell 9 ---
import numpy as np

x_train_transformed= np.concatenate(
    (x_train_age, x_train_gender_city, x_train_cough, x_train_fever),
    axis=1
)

x_test_transformed = np.concatenate(
    (x_test_age, x_test_gender_city, x_test_cough, x_test_fever),
    axis=1
)

x_train_transformed.shape

# --- Cell 10 ---
from sklearn.compose import ColumnTransformer

cl = ColumnTransformer(transformers=[
    ("tr1",SimpleImputer(),["fever"]),
    ("tr2",OrdinalEncoder(categories=[["Mild","Strong"],]),["cough"]),
    ("tr3",OneHotEncoder(drop="first",sparse_output=False),["gender","city"])],
    remainder="passthrough"
)

# --- Cell 11 ---
cl.fit_transform(x_train).shape

# --- Cell 12 ---
# First, let's actually transform both our training and testing data using the pipeline we built
x_train_transformed = cl.fit_transform(x_train)
x_test_transformed = cl.transform(x_test)

# --- Cell 13 ---
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# Initialize the model
model = DecisionTreeClassifier(random_state=42)

# Train (fit) the model on the transformed training data
model.fit(x_train_transformed, y_train)

# --- Cell 14 ---
# Make predictions on the test data that the model hasn't seen before
y_pred = model.predict(x_test_transformed)

# Let's look at the first few predictions vs the actual values
print("Predictions:", y_pred[:5])
print("Actuals:    ", y_test[:5].values)

# --- Cell 15 ---
# Calculate and print the accuracy score
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy * 100:.2f}%")
```
